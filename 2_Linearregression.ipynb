```python
import pandas as pd
```

```python
path='/content/Salary_Data.csv'
path
```

```python
#Load dataset
sal=pd.read_csv(path)
sal[:2]
```

```python
#Prepare data
X=sal[['YearsExperience']] #ML algorithms always takes an input in 2D

y=sal['Salary']
```

```python
#data analysis
sal['Salary'].values
```

```python
sal['Salary']
```

```python
#check relationship of the data
import matplotlib.pyplot as plt
plt.scatter(X,y,color='red',marker= '*')
```

```python
#import algorithm--> scikit learn -->sklearn

from sklearn.linear_model import LinearRegression
```

```python
#As it is class we need to create a constructor-->object
model=LinearRegression() #LinearRegression() is class calling
```

```python
model #object
```

```python
#Training --> using fit() method
model.fit(X,y)
```

```python
#When i say traning:-means it is calculating the slope and intercept which is required for the line
#formation the line equation
```

```python
#Testing using predict method
import warnings
warnings.filterwarnings('ignore')
model.predict([[0]])
```

```python
model.predict([[7.5]])
```

```python
sal.describe().T
```

#Equation:- Y=mx+c

```python
#m-slope
m = model.coef_
m
```

```python
# c-intercept
c = model.intercept_
c
```

```python
m*7.5 +c
```

```python
model.predict([[11]])
```

```python
m*11+c #formula
```

```python
test_data = [[0],[4.5],[6],[11]]
```

```python
y_pred = model.predict(test_data)
y_pred
```

```python
#actual data points
plt.scatter(X,y,color='red',marker='*')

# test line/line of prediction
plt.plot(test_data,y_pred,color='k')
```
